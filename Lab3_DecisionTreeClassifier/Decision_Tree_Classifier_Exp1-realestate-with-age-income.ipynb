```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

```python
import kagglehub

path = kagglehub.dataset_download("baydarenes/customer-attributes-and-purchase-propensity")
csv_path=path+"/Customer_Attributes_and_Purchase_Propensity.csv"
data=pd.read_csv(csv_path)
data.head()
```

```python
x = data[['Age', 'Income']]
y = data['Purchased']
model = DecisionTreeClassifier(criterion='gini', max_depth=2)
model.fit(x, y)
```

*Chose a decision tree based representation because it  shows how data is split in every step explaining the decision-making process as described similarly in the question*

```python
plt.figure(figsize=(10, 6))
plot_tree(model, feature_names=['Age', 'Income'], class_names=['No', 'Yes'], filled=True)
plt.show()
```
